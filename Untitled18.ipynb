```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

from IPython.display import display

# Load customer dataset
url = "https://raw.githubusercontent.com/marceloreis/Mall_Customers/master/Mall_Customers.csv"

try:
    df = pd.read_csv(url)
    print("Dataset loaded successfully!")

except Exception:
    print("Download failed. Creating sample dataset.")

    np.random.seed(42)

    df = pd.DataFrame({
        "CustomerID": range(1, 301),
        "Gender": np.random.choice(["Male", "Female"], 300),
        "Age": np.random.randint(18, 70, 300),
        "Annual Income (k$)": np.random.randint(15, 140, 300),
        "Spending Score (1-100)": np.random.randint(1, 100, 300)
    })

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(df["Annual Income (k$)"], kde=True)
plt.title("Annual Income Distribution")

plt.subplot(1, 2, 2)
sns.histplot(df["Spending Score (1-100)"], kde=True)
plt.title("Spending Score Distribution")

plt.tight_layout()
plt.show()
```

```python
# Select features
X = df[[
    "Annual Income (k$)",
    "Spending Score (1-100)"
]]

# Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Elbow Method
inertia = []

for k in range(1, 11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)
    inertia.append(model.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inertia, marker="o")
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.grid(True)
plt.show()
```

```python
# Create K-Means model
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

# Predict customer clusters
df["Cluster"] = kmeans.fit_predict(X_scaled)

print("Customer clustering completed!")

display(df.head())
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    hue="Cluster",
    palette="viridis",
    s=100
)

plt.title("MarketMind Customer Segmentation")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score")
plt.grid(True)
plt.show()
```

```python
# Calculate average values for each cluster
profile = df.groupby("Cluster").agg({
    "Annual Income (k$)": "mean",
    "Spending Score (1-100)": "mean",
    "CustomerID": "count"
}).rename(columns={"CustomerID": "Customer Count"})

# Name customer segments
def segment_name(row):
    income = row["Annual Income (k$)"]
    spending = row["Spending Score (1-100)"]

    if income >= 65 and spending >= 60:
        return "Target Customers"

    elif income >= 65 and spending < 40:
        return "Careful High-Income"

    elif income < 40 and spending >= 60:
        return "High-Spending Budget"

    elif income < 40 and spending < 40:
        return "Low-Value Customers"

    else:
        return "Average Customers"

profile["Segment Name"] = profile.apply(
    segment_name,
    axis=1
)

# Add segment names to customer data
segment_mapping = profile["Segment Name"].to_dict()

df["Segment Name"] = df["Cluster"].map(segment_mapping)

display(profile)
```

```python
fig = px.scatter(
    df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    color="Segment Name",
    hover_data=["CustomerID", "Age", "Gender"],
    title="MarketMind – Customer Segmentation Dashboard",
    template="plotly_white"
)

fig.update_traces(marker=dict(size=12))

fig.update_layout(
    width=1000,
    height=600,
    legend_title="Customer Segments"
)

fig.show()
```

```python
# Count customers in each segment
segment_counts = df["Segment Name"].value_counts()

plt.figure(figsize=(9, 5))

sns.barplot(
    x=segment_counts.index,
    y=segment_counts.values
)

plt.title("Number of Customers in Each Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
recommendations = {
    "Target Customers":
        "Offer loyalty rewards, premium products and exclusive deals.",

    "Careful High-Income":
        "Provide premium quality products and personalized offers.",

    "High-Spending Budget":
        "Offer discounts, affordable bundles and value packs.",

    "Low-Value Customers":
        "Use special promotions to increase customer engagement.",

    "Average Customers":
        "Encourage repeat purchases through membership programs."
}

for segment in df["Segment Name"].unique():
    print("\nCustomer Segment:", segment)
    print("Recommendation:", recommendations.get(segment))
```

```python
score = silhouette_score(X_scaled, df["Cluster"])

print("Silhouette Score:", round(score, 3))
```
